# easySLM - Inference & Prediction Playground

This notebook loads a fine-tuned model, runs batch predictions on the validation set, and provides an interactive cell for manual single-text inference showing both raw and decoded outputs.

```python
from pathlib import Path

from src.slm import SLMBase

# 1. Load fine-tuned model checkpoint (select any model in models/finetuned/)
CHECKPOINT = "./models/finetuned/Qwen2.5-0.5B_json_20261105_165938"
model = SLMBase.from_checkpoint(CHECKPOINT)
```

```python
# 2. Predict on validation set
val_lines = Path("data/movie_plots/val.txt").read_text(encoding="utf-8").splitlines()
results = model.predict_lines(val_lines)

print(f"Total validation samples evaluated: {len(results)}\n")
for r in results[:5]:  # Display first 5 samples
    print(f"Text:         {r['text'][:75]}...")
    print(f"Prediction:   {r['predictions']}")
    print(f"Ground Truth: {r.get('ground_truth')}\n")
```

```python
# 3. Interactive single-text inference (Raw & Decoded output)
custom_text = "no mvie"

res = model.predict_texts([custom_text])[0]
raw_output = res.get("probabilities") if model.mode == "cls" else res.get("raw_output")

print("--- INPUT TEXT ---")
print(custom_text)
print("\n--- RAW OUTPUT ---")
print(raw_output)
print("\n--- DECODED OUTPUT ---")
print(res["predictions"])
```
